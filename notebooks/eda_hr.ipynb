```python
import pandas as pd

# Carregar os ficheiros CSV que exportámos do FreeSQL
df_salarios = pd.read_csv('../data/query_01.csv')
df_regioes = pd.read_csv('../data/query_02.csv')

# Exibir as primeiras 5 linhas da tabela de salários para validar a importação
display(df_salarios.head())
```

```python
# --- 1. Inspeção da Estrutura dos Dados ---
print("--- Dimensões da Tabela de Salários (Linhas, Colunas) ---")
print(df_salarios.shape)

print("\n--- Informações dos Tipos de Dados ---")
df_salarios.info()

# --- 2. Verificação de Qualidade ---
print("\n--- Contagem de Valores Nulos por Coluna ---")
print(df_salarios.isnull().sum())

print("\n--- Contagem de Linhas Duplicadas ---")
print(f"Duplicados: {df_salarios.duplicated().sum()}")
```

```python
# --- 3. Estatísticas Descritivas dos Salários ---
print("--- Resumo Estatístico Básico ---")
display(df_salarios['SALARY'].describe())

print("\n--- Métricas Individuais Específicas ---")
media = df_salarios['SALARY'].mean()
mediana = df_salarios['SALARY'].median()
minimo = df_salarios['SALARY'].min()
maximo = df_salarios['SALARY'].max()
desvio_padrao = df_salarios['SALARY'].std()

print(f"Média de Salários: {media:.2f}")
print(f"Mediana de Salários: {mediana:.2f}")
print(f"Salário Mínimo: {minimo:.2f}")
print(f"Salário Máximo: {maximo:.2f}")
print(f"Desvio Padrão: {desvio_padrao:.2f}")
```

```python
import matplotlib.pyplot as plt

# Configurar o tamanho da figura para acomodar dois gráficos lado a lado
plt.figure(figsize=(14, 5))

# --- Gráfico 1: Histograma ---
plt.subplot(1, 2, 1)
df_salarios['SALARY'].plot(kind='hist', bins=15, color='#4C72B0', edgecolor='black')
plt.title('Distribuição de Salários (Histograma)')
plt.xlabel('Salário ($)')
plt.ylabel('Número de Funcionários')

# --- Gráfico 2: Boxplot ---
plt.subplot(1, 2, 2)
df_salarios['SALARY'].plot(kind='box', color='green', vert=False)
plt.title('Distribuição de Salários (Boxplot)')
plt.xlabel('Salário ($)')

# Ajustar o layout e exibir
plt.tight_layout()
plt.show()
```
